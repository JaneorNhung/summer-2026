# VECTORIZATION
Instead of telling Python what to do with each element, you tell Numpy the operation for the whole array
### 1. Example (double the array):

```python
# Python
numbers = [1,2,3,4,5]
result = []
for num in numbers:
    result.append(num*2)
print(result)
# numpy
import numpy as np
a = np.array([1,2,3,4,5])
a*2
```

### 2. Numpy functions

```python
np.sqrt(a) # square root
np.abs(a) # |a|
np.exp(a) # e^a
np.log(a) # ln(a)
np.log2(a) # log2(a)
```

# RANDOM
### 1. np.random.rand()

```python
# generates random float number btw 0 and 1
np.random.rand()
```

```python
# multiple numbers
np.random.rand(5)
```

```python
# matrix
np.random.rand(3,2)
```

### 2. np.random.randint()
generates random integers

```python
#one integer
np.random.randint(1,10)
```

```python
#vector
np.random.randint(1,10,size=(3,2))
```

### 4. np.random.seed()
will always give the same sequence after setting that seed.

```python
np.random.seed(42)
print(np.random.rand(3))
```
