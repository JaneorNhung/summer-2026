```python
# Exercise 1
# Write a function that takes a number and returns its square.
def square(x):
    return x**2
print(square(5))
```

```python
# Exercise 2
# Write a function that takes two numbers and returns their sum.
def add(x,y):
    return x+y
print(add(3,4))
```

```python
# Exercise 3
# Write a function that takes a person's age and returns:
# `"child"` if under 13
# `"teenager"` if 13–17
# `"adult"` if 18+
def age_classify(x):
    if x<13:
        return "child"
    elif x<=17:
        return "teenager"
    else:
        return "adult"
```

```python
# Exercise 4
# Write a function that takes a list of numbers and returns the largest number.
def largest(numbers):
    largest_number = numbers[0]
    for number in numbers:
        if number > largest_number:
            largest_number = number
    return largest_number
print(largest(numbers))
```

```python
#Add an element
numbers.append(7)
print(numbers)
```

```python
# Loop through a list:
for num in numbers:
    print(num)
```

```python
# Return the smallest number of a list
def smallest(numbers):
    min = numbers[0]
    for num in numbers:
        if num < min:
            min = num
    return min
print(smallest(numbers))
```

```python
# Return average value of a list
def average(numbers):
    total = 0
    for num in numbers:
        total += num
    return total/len(numbers)
print(average(numbers))
```

```python
# Return the list of even numbers
def even_list(numbers):
    even = []
    for num in numbers:
        if num%2==0:
            even.append(num)
    return even
print(even_list(numbers))
```
