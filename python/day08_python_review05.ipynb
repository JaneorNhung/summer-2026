# TUPLE
An immutable list

```python
# Declare
point = (3,5) 
```

### 1. Indexing

```python
point[0]
point[-1]
```

### 2. Slicing

```python
numbers = (1,2,7,4,57,3)
numbers[1:4]
```

### 3. Unpacking

```python
point = (3,5)
x, y = point
x
y
```

```python
# same with string
student = ("Jane", 18, "DS")
name, age, major = student
age
```
