###### PRACTICE FOR BOOLEAN MASKING

```python
import numpy as np
a= np.arange(1,13).reshape(3,4)
print(a)
```

```python
#1 Take out numbers > 7
a[a>7]
```

```python
#2 Take out all even numbers
a[a%2==0]
```

```python
#3 Take out all numbers belong to [4,9]
a[(a>=4) & (a<=9)]
```

```python
#4 Replace all numbers that > 8 with 100
#C1
b = np.where(a>8, 100, a) 
print(b)
```

###### /Keep the other values by 'a'/

```python
#C2
a[a>8] = 100
print(a)
```

```python
#5 Find indexes of all numbers that are >8 using np.where()
np.where(a>8)
```

```python
#6 Find indices of non-zero elements from given list
arr = np.array([1,2,0,0,4,0])
np.where(arr!=0)
```

```python
#7 Given an 1D array, negate all elements which are between 3 and 8 in place
arr[(arr>=3) & (arr<=8)] *= -1
print((arr))
```

```python
#8 How to find common values between 2 arrays?
a= np.array([1,2,3,6])
b= np.array([3,4,5,6])
np.intersect1d(a,b)
```

```python
#9 Create random vector of size 10 and replace maximum value by 0
array = np.random.randint(1,100, size = 10)
array[np.argmax(array)] = 0
print(array)
```
