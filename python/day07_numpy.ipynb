# COMBINING ARRAYS
### 1. np.concatenate()

```python
a = np.array([[2,4],[6,8]])
b = np.array([[3,5],[7,9]])
np.concatenate((a,b), axis = 0) #or no axis, same for 2*2 arr
```

```python
np.concatenate((a,b), axis = 1)
```

axis = 0: Adding rows-    (2,2) + (2,2) = (4,2)
axis = 1: Adding columns- (2,2) +(2,2) = (2,4)

### 2. np.vstack()
vstack = vertical stack

```python
np.vstack((a,b)) #convenient version of np.concatenate((a,b), axis = 0)
```

### 3. np.hstack()
hstack = horizontal stack

```python
np.hstack((a,b)) #convenient version of np.concatenate((a,b), axis = 1)
```

### 4. np.split()

```python
a = np.arange(1,7)
np.split(a,3)
```

```python
b= np.arange(1,13).reshape(3,4)
np.split(b, 2, axis = 1)
```

# tile()
##### Repeat an array/pattern

```python

# 1D
a = np.array([1,2,3])
np.tile(a,2)
```

```python
# 2D
a = np.array([[1,2,3],[2,3,4]])
np.tile(a,(2,3))
# rows repeat
# columns repeat
```

# BROADCASTING

```python
a = np.array([1,6,7])
a+10
```

```python
a= np.array([[1,2,3], [4,5,6]])
b = np.array([10,20,30])
a+b
```

#### When does it work?
2 dimensions are compatible if:
+ they are equal
+ one of them is 1
+ one array doesnt have that dimension
Ex1: (2,3) + (3) --> compatible (3=3)
Ex2: (2,3) + (2) --> incompatible (2!=3- compare from the right)

# MEAN (average)

```python
scores = np.array([
    [80, 70, 90],
    [60, 85, 75],
    [95, 90, 100]
])
np.mean(scores, axis = 0)
np.mean(scores, axis = 1)
```

# MEDIAN (The middle value after sorting)
Median is less affected by extreme outliers

```python
a = np.array([1,2,3,100])
np.median(a)
```

# STANDARD DEVIATION np.std()
measure how spread out the data is

```python
a = np.array([1,2,3,4,5])
np.std(a)
```

Meaning:
+ small std: values are close together
+ large std: values are more spread out

# keepdims

```python
a = np.array([
    [10, 20, 30],
    [40, 50, 60]
])
row_mean = np.mean(a, axis=1, keepdims = True)
a - row_mean
```

it is useful in broadcasting 'cause it keeps the reduced dimension as size 1
