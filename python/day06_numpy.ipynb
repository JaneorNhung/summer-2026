# BOOLEAN COMPARISON

```python
import numpy as np
a = np.array([[6,4,9],[2,5,7]])
```

```python
# Boolean mask
a>5
```

```python
# Use mask to filter
a[a>5]
a[a!=5]
```

##### For multiple conditions: 
###### AND &
###### OR |
###### NOT ~

```python
a[(a>3) & (a<7)]
```

###### np.where

```python
# Find position
a= np.array([[9,4,7],[2,6,3]])
np.where(a>4) #1st array returns row indices, 2nd returns column indices
```

```python
# Conditional replacement
b= np.where(a>5, 100, 0) #a>5 replace with 100, a<=5 replace with 0
print(b)
```

##### Note: np.where does not adjust the original array, but an alternative

```python
# Find position of extreme value
np.argmax(a)
np.argmin(a, axis = 1)
```
