# DICTIONARY

```python
student = {
    "name": "Jane",
    "age": 19
}
```

### A. Access (using a key)

```python
student["name"]
student["age"]
```

list->       index-> value
dictionary-> key->   value

### B. Add/Update

```python
student["major"] = "Data Science"
student["age"] = 18
```

### C. .get()

```python
student.get("age")
```

Difference between students.get() and student[]:
If student[] doesnt exist     -> error
If students.get() doesnt exist-> None

### D. For in dictionary

```python
for key, value in student.items():
    print(key, value)
```

all keys       -> student.keys()
all values     -> student.values()
key-value pairs-> student.items()
