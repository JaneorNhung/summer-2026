**  LINEAR ALGEBRA **

```python
import numpy as np
a= np.ones((2,3))
print(a)
b= np.full((3,2), 2)
print(b)
np.matmul(a,b)
```

```python
#Find the determinant
c= np.identity(3)
np.linalg.det(c)
```

```python
#SVD
np.linalg.svd(c)
```

```python
#Trace
np.trace(c)
```

```python
#Eigenvalues
np.linalg.eig(c)
```

```python
#Matrix norm
np.linalg.norm(c)
```

```python
#Inverse
np.linalg.inv(c)
```

```python
#MAX
np.max(arr, axis = 0)
```
