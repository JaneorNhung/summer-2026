### Functions + Lists

```python
def sum_list(numbers):
    total = 0
    for num in numbers:
        total += num
    return total
numbers = [2,3,7,5,1]
print(sum_list(numbers))
```

```python
def reverse_list(numbers):
    reverse = []
    for num in numbers:
        reverse.insert(0,num)
    return reverse
print(reverse_list(numbers))
```

```python
def count_positive(numbers):
    count = 0
    for num in numbers:
        if num>0:
            count = count + 1
    return count
print(count_positive(numbers))
```

```python
def find_duplicates(numbers):
    duplicates = []
    for i in range(len(numbers)):
        for j in range(i+1, len(numbers)):
            if numbers[i] == numbers[j]:
                if numbers[i] not in duplicates:
                    duplicates.append(numbers[i])
    return duplicates
numbers = [1,2,7,4,8,2,4,8,2]
print(find_duplicates(numbers))
```

```python
def remove_duplicates(numbers):
    result = []
    for number in numbers:
        if number not in result:
            result.append(number)
    return result
print(remove_duplicates(numbers))
        
```

#### MORE ABOUT LIST

```python
# append(value) add an element at the end of the list
numbers = [1,2,3,4,5]
numbers.append(6)
print(numbers)
```

```python
# remove(value) remove a value (the first value only) out of the list
numbers.remove(4)
print(numbers)
```

```python
# pop(index) remove a value depending on its index & return the deleted value
x = numbers.pop(2)
print(numbers)
print(x)
numbers.pop() #remove the last element
```

```python
#sort() sort list
numbers.sort()
print(numbers)
```

```python
#sort(reverse=True) sort descending list
numbers.sort(reverse= True)
print(numbers)
```

```python
#Numbers of elements
print(len(numbers))
```
