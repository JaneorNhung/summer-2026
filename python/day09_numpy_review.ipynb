```python
#3 enumerate() & np.ndenumerate()
# enumerate() for 1d array only
numbers = [10,20,30]
for i, value in enumerate(numbers):
    print(i, value)
# np.ndenumerate for n dimensional arrays
a = np.array([
    [10,20],
    [30,40]
])
for index, value in np.ndenumerate(a):
    print(index, value)
```

```python
#3.1 Print index, value
a = np.array([
    [5, 8],
    [2, 9]
])
for index, value in np.ndenumerate(a):
    print(index,value)
```

```python
#3.2 Find and print (index,value) with value > 5
a = np.array([
    [5, 8],
    [2, 9]
])
for index, value in np.ndenumerate(a):
    if value>5:
        print(index,value)
```

```python
#4 Sort rows according to the 2nd columnm
a = np.array([
    [3, 8, 1],
    [5, 2, 9],
    [4, 7, 6]
])
order = np.argsort(a[:,1]) #sort the 2nd column
a[[order]] #fancy indexing 
```

Pattern: 
order = np.argsort(a[row,col])
a[[order]]

```python
#4.2 Sort rows theo col 1
a = np.array([
    [1, 90, 5],
    [2, 70, 8],
    [3, 80, 2]
])
order = np.argsort(a[:,0])
a[[order]]
```

```python
#6 Check if there is any full zero column
a = np.array([
    [1, 0, 3],
    [2, 0, 4],
    [5, 0, 6]
])
np.all(a==0, axis = 0)
```

Pattern:
+ np.all() AND
np.all(condition, axis (if any))
Ex: np.all(age>5)
Ex: np.all(a == b)
+ np.any() OR



```python
#6.1 Find columns that are full-zero
a = np.array([
    [0, 1, 0],
    [0, 2, 0],
    [0, 3, 0]
])
np.all(a==0, axis = 0)
```

```python
#6.2 Find columns/rows that are full-zeros
a = np.array([
    [1, 0, 3],
    [0, 0, 0],
    [2, 0, 5]
])
np.all(a==0, axis = 0)
np.all(a==0, axis = 1)
```
