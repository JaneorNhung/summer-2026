##DAY 2: INITIALIZING DIFFERENT TYPES OF ARRAYS

```python
#All 0s matrix
import numpy as np
np.zeros((2,3))
```

```python
#All 1s matrix
np.ones((3,2,2), dtype=np.int32)
```

```python
#Any other number
np.full((2,2), 99)
```

```python
a = np.array([[1,2,3], [4,5,6]]) 
```

```python
#Any other number (full-like)
np.full_like(a, 4)
```

```python
#Random decimal number
np.random.rand(4,2)
```

```python
#Random decimal number based on a shape
np.random.random_sample(a.shape)
```

```python
#Random integer numbers
np.random.randint(7, size=(2,2))
np.random.randint(-2,6,(3,2))
# (start, stop (exclusive), size)
```

```python
#The Identity matrix
np.identity(5)
```

```python
#Repeat an array
arr = np.array([[1,2,3], [2,3,4]])
r1 = np.repeat(arr,2,axis =1)
print(r1)
```

```python
#Print out a specific matrix
output = np.ones((5,5))
print(output)
z = np.zeros((3,3))
z[1,1] = 9
print(z)
output[1:4, 1:4] = z
print(output)
```

*Be careful when copying array!*

```python
a = np.array([1,2,3])
b = a.copy()
b[0] = 100
print(a)
# if 
# b=a
# b[0] = 100
# a= [100,2,3] = b
```
