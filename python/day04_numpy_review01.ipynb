# C. Indexing & Slicing

```python
a = np.arange(1,13).reshape(3,4)
```

```python
# First row
a[0,:]
```

```python
# Second column
a[:,1]
```

```python
# Last row
a[-1,:]
```

```python
# Last column
a[:,-1]
```

```python
# Middle 2 columns
a[:, 1:3]
#start:end:step, end is exclusive
```

```python
# First 2 rows
a[0:1, :]
```

```python
# Every other column
a[:, ::2]
```

```python
# Bottom-right element
a[-1, -1]
```
