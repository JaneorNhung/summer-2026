# SET
Set doesn't have duplicates and indexing

### 1. Add

```python
s = {1,2,3}
s.add(4)
s
```

### 2. Remove/Discard

```python
# Remove
s.remove(2)
print(s)
# Discard
s.discard(5)
print(s)
```

For a non-exist element, 
+ remove: error
+ discard: no error

### 3. Union

```python
# Phep hop
a = {"An", "Binh", "Cuong"}
b= {"Binh", "Cuong", "Dung"}
a.union(b)
```

### 4. Intersection

```python
#Phep giao
a.intersection(b)
```

### 5. Difference

```python
# Phep hieu a\b- available at a but not b
a.difference(b)
```

### 6. Transform a list to a set

```python
numbers = [1,2,3,3,4,4,5]
set(numbers)
```

### 7. Check whether b is in a

```python
b.issubset(a) 
#True: in
#False: not in
```

### 8. Practice

```python
# 1
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
# All elements belong to a & b
a.union(b)
# Elements in common
a.intersection(b)
# Elements belong to a only
a.difference(b)
```

```python
# 2
numbers = [1, 2, 2, 3, 4, 4, 5, 5, 5]
set_numbers = set(numbers)
print(set_numbers)
```

```python
# 3
a = {1, 2, 3, 4, 5}
b = {2, 4}
b.issubset(a)
```

```python
# 4
students_a = {"An", "Binh", "Chi", "Dung"}
students_b = {"Chi", "Dung", "Hoa", "Lan"}
# students studying in both class
students_a.intersection(students_b)
# students studying in a only
students_a.difference(students_b)
# all students
students_a.union(students_b)
# students studying in b only
students_b.difference(students_a)
```
