### Mini Challenge 
Write a program that:
Stores your monthly income.
Stores your monthly expenses.
Calculates your savings.
Prints:
Income: xxxx
Expenses: xxxx
Savings: xxxx

```python
income = float(input("Your monthly income: "))
expenses = float(input("Your monthly expenses: "))
savings = income - expenses
print("Monthly Summary")
print(f"Income: {income}")
print(f"Expenses: {expenses}")
print(f"Savings: {savings}")
```
