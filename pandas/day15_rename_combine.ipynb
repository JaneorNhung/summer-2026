# 2. Combining
### 2.1 concat()
combine multiple series/dataframes along rows/cols (put A underneath B)

```python
pd.concat([df1, df2,...], axis = 0)
# axis = 0: rows
# axis = 1: cols
```

```python
mon = pd.Dataframe({
    "product": ['computer', 'mouse'],
    "sales": [240, 410]
})
tue = pd.Dataframe({
    " product": ['phone', 'headphone'],
    "sales": [400, 230]
})
result = pd.concat([mon, tue], ignore_index = True)
```
