### 10. .rank(method = ' ', ascending = True/False)
Ex: [7,9,9,10]
- method = 'average': take the average of the duplicated ranks. [1, 2.5, 2.5, 4]
- method = 'min': take the maximum rank of the duplicated ranks [1,2,2,4]
- method = 'max': take the minimum rank of the duplicated ranks [1,3,3,4]
- method = 'first': no duplicated ranks after sorted, if duplicated-> take the ones that appears first [1,2,3,4]
- method = 'dense': same with min, but no skipping rank [1,2,2,3]
NOTE: the highest = 1st rank--> ascending = False
