# Groupwise Analysis
### 1. value_counts
Group identical values, outputing how many times each value appears in that column

```python
reviews['column_name'].value_counts()

#Ex: how many bottles of wine each country produced
reviews.country.value_counts()
```

Features of value_counts:
+ sorted from Highest to Lowest
+ ignore missing values (NaN)
+ if want to know the percentage each category makes up

```python
reviews['country'].value_counts(normalize = True)
```

### 2. groupby()
#### What is it for?
+ SPLIT: split a dataframe into smaller, separate mini-tables (buckets) based on the category chosen
+ APPLY: apply a function to each bucket independently
+ COMBINE: glue all those individual answers back into a table

#### When to use?
per, for each

```python
# divide rows into groups based on 'country'
reviews.groupby('country')
```

```python
# average price per country
reviews.groupby('country').price.mean()
```

result can be:
country
France    50
US        33.3

```python
# the highest score for each winery?
reviews.groupby('winery').points.max()
# how many wines are there per province?
reviews.groupby('province').title.count()
```

```python
# divide data based on 'country' and 'province'
reviews.groupby(['country', 'province'])
# each pair (country, province) is a group
```

.apply() basically means: Run this function on each group

```python
def hello(df):
    return "hello"
reviews.groupby('winery').apply(hello)
```

lambda is a short way to write a function. instead of

```python
def square(x):
    return x ** 2
# you can write
lambda x: x ** 2
```

```python
# Take out the first review title of each winery
reviews.groupby('winery').apply(lambda df: df.title.iloc[0])
```

### 3. agg() - aggregate
Calculate summaries for each group the give the results as a table

```python
reviews.groupby(['country']).price.agg([len, min, max])
```

| country | len | min | max |
| ------- | --: | --: | --: |
| France  |   2 |  40 |  60 |
| US      |   3 |  20 |  50 |
